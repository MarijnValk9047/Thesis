# 03 SARIMA

This notebook reviews the first SARIMA benchmark.

Model summary:
- SARIMA extends ARIMA with an explicit seasonal structure
- for hourly electricity prices, that seasonal structure is used to capture repeating within-day patterns
- in this workstream SARIMA is still a univariate price-only benchmark, so the comparison stays methodologically clean

```python
from pathlib import Path
import json
import pandas as pd

run_root = Path('data/02_Forecasting/01_DA_prices/hourly_da/runs')
latest_run = sorted(run_root.glob('*_sarima_benchmark'))[-1]
latest_run
```

```python
metrics_overall = pd.read_csv(latest_run / 'metrics_overall.csv')
metrics_by_lead_day = pd.read_csv(latest_run / 'metrics_by_lead_day.csv')
timing_summary = pd.read_csv(latest_run / 'origin_timing_summary.csv')
dm_results = pd.read_csv(latest_run / 'diebold_mariano_results.csv')
official_naive = json.loads((latest_run / 'official_naive_reference.json').read_text(encoding='utf-8'))

display(metrics_overall)
display(metrics_by_lead_day[metrics_by_lead_day['model'] == 'sarima'])
display(timing_summary)
display(dm_results[dm_results['challenger_model'] == 'sarima'])
official_naive
```

Interpretation notes:

- SARIMA uses the same rolling-origin schedule, split policy, and output schema as the naive and ARIMA runs.
- The seasonal component makes the model heavier, which is why the timing monitor is especially relevant here.
- Any later model-comparison notebook can now combine naive, ARIMA, and SARIMA results without reworking the evaluation logic.
