# 06 Model Comparison

This notebook consolidates all implemented hourly DA models into one comparison layer.

Included models:
- previous-week naive
- previous-year naive
- ARIMA
- SARIMA
- LEAR `FS1` and `FS2`
- XGBoost `FS1` and `FS2`

Comparability rules enforced here:
- same splits
- same daily rolling-origin logic
- same `08:00` D-1 forecast origin
- same `D` through `D+4` horizon
- same long-format storage schema
- same core metrics and Diebold-Mariano testing

```python
from pathlib import Path
import json
import pandas as pd

run_root = Path('data/02_Forecasting/01_DA_prices/hourly_da/runs')
latest_run = sorted(run_root.glob('*_model_comparison'))[-1]
latest_run
```

```python
metrics_overall = pd.read_csv(latest_run / 'metrics_overall.csv')
metrics_by_lead_day = pd.read_csv(latest_run / 'metrics_by_lead_day.csv')
dm_results = pd.read_csv(latest_run / 'diebold_mariano_results.csv')
timing_summary = pd.read_csv(latest_run / 'origin_timing_summary.csv')
source_runs = json.loads((latest_run / 'source_runs.json').read_text(encoding='utf-8'))

display(metrics_overall)
display(metrics_by_lead_day.head(18))
display(timing_summary)
source_runs
```

How to read the results:

- `validation` is used for benchmark choice and model comparison.
- `test` is held back for the final out-of-sample assessment.
- `rmae_vs_official_naive` is the main relative metric for this workstream.
- Diebold-Mariano results are directional here: a negative statistic means the challenger has lower loss than the benchmark on average.
