# 01 Naive Benchmark

This notebook evaluates the two required naive baselines:
- previous-week naive
- previous-year naive

Why both are kept:
- both are stored and reported
- the official naive benchmark is selected only from validation performance
- relative metrics such as rMAE use that official benchmark afterwards

Methodological detail:
- the naive models are aligned to local delivery time, not just a fixed UTC shift
- history availability is based on the `known_at` assumption, so the full DA curve for day `D-1` is available at the `D-1 08:00` forecast origin
- this matters on DST transition days

```python
from pathlib import Path
import json
import pandas as pd

run_root = Path('data/02_Forecasting/01_DA_prices/hourly_da/runs')
latest_run = sorted(run_root.glob('*_naive_benchmark'))[-1]
latest_run
```

```python
metrics_overall = pd.read_csv(latest_run / 'metrics_overall.csv')
metrics_by_lead_day = pd.read_csv(latest_run / 'metrics_by_lead_day.csv')
official_naive = json.loads((latest_run / 'official_naive_reference.json').read_text(encoding='utf-8'))

display(metrics_overall)
display(metrics_by_lead_day.head(12))
official_naive
```

How to read the main outputs:

- `metrics_overall.csv` reports validation and test performance over all scored origin-target pairs.
- `metrics_by_lead_day.csv` breaks the same evaluation down into `D`, `D+1`, ..., `D+4`.
- `rmae_vs_official_naive = 1.0` for the chosen official naive benchmark by construction.
- Lower MAE, RMSE, and rMAE are better. Bias near zero is preferable.
