# 07 Visual Case Weeks

This notebook inspects objective case-study weeks from the **test split only**.

Selection logic:
- full Monday-Sunday ISO weeks inside the test period are considered
- weekly actual-price features are computed first
- a typical winter week is the winter candidate closest to the winter centroid in standardized feature space
- a typical summer week is the summer candidate closest to the summer centroid in standardized feature space
- the high-volatility week is the eligible week with the highest weekly standard deviation

Plotting note:
- the visual overlays use stitched `D`-only operational day-ahead forecasts so each target hour has one forecast per model

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import Image, display

run_root = Path('data/02_Forecasting/01_DA_prices/hourly_da/runs')
latest_run = sorted(run_root.glob('*_visual_case_weeks'))[-1]
latest_run
```

```python
weekly_features = pd.read_csv(latest_run / 'weekly_features.csv')
candidate_weeks = pd.read_csv(latest_run / 'candidate_weeks.csv')
selected_weeks = pd.read_csv(latest_run / 'selected_weeks.csv')
week_metrics = pd.read_csv(latest_run / 'week_metrics.csv')
run_summary = json.loads((latest_run / 'run_summary.json').read_text(encoding='utf-8'))

display(selected_weeks)
display(candidate_weeks)
display(week_metrics)
run_summary
```

How to use these outputs:

- `candidate_weeks.csv` shows the top ranked weeks in each category.
- `selected_weeks.csv` contains the final representative weeks.
- `week_metrics.csv` compares per-week errors across all implemented models.
- the plots below provide the thesis-ready visual comparison subset.

```python
for category in selected_weeks['category']:
    display(Image(filename=latest_run / 'plots' / f'{category}_actual_only.png'))
    display(Image(filename=latest_run / 'plots' / f'{category}_actual_vs_all_models.png'))
```
