# 02 ARIMA

This notebook reviews the first ARIMA benchmark.

Model summary:
- ARIMA is a univariate time-series model
- it uses only the historical DA price series
- in this thesis pipeline it is the closest comparable price-only benchmark against later richer feature-based models

Current benchmark setup:
- shared daily rolling-origin evaluation
- shared UTC storage and metric definitions
- ARIMA is compared against both naive baselines, with rMAE reported relative to the validation-selected official naive benchmark

```python
from pathlib import Path
import json
import pandas as pd

run_root = Path('data/02_Forecasting/01_DA_prices/hourly_da/runs')
latest_run = sorted(run_root.glob('*_arima_benchmark'))[-1]
latest_run
```

```python
metrics_overall = pd.read_csv(latest_run / 'metrics_overall.csv')
metrics_by_lead_day = pd.read_csv(latest_run / 'metrics_by_lead_day.csv')
timing_summary = pd.read_csv(latest_run / 'origin_timing_summary.csv')
dm_results = pd.read_csv(latest_run / 'diebold_mariano_results.csv')
official_naive = json.loads((latest_run / 'official_naive_reference.json').read_text(encoding='utf-8'))

display(metrics_overall)
display(metrics_by_lead_day[metrics_by_lead_day['model'] == 'arima'])
display(timing_summary)
display(dm_results[dm_results['challenger_model'] == 'arima'])
official_naive
```

Interpretation notes:

- ARIMA is intentionally feature-light here. That keeps the benchmark transparent and directly comparable in the shared walk-forward setup.
- This run uses a practical rolling history window and iteration cap for feasibility. It is a first benchmark, not a full tuning study.
- Timing outputs show whether daily refitting is operationally manageable.
